```python
import numpy as np

#AND gate inputs and expected outputs
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])
```

```python
#step activation function
def activate(output):
    if output >= 0:
        return 1
    else:
        return 0
```

```python
def perceptron_and_gate(X, y, learning_rate=0.1, epochs=10):
    # Initialize weights and bias
    weights = np.array([0.0, 0.0])
    bias = 0.0

    print(f"Initial Weights: {weights}, Bias: {bias}\n")

    for epoch in range(epochs):
        print(f"Epoch {epoch + 1}/{epochs}")
        total_error = 0
        for i in range(len(X)):
            output = np.dot(X[i], weights) + bias
            prediction = activate(output)

            error = y[i] - prediction
            total_error += abs(error)

            # Update weights and bias
            weights = weights + learning_rate * error * X[i]
            bias = bias + learning_rate * error

            print(f"Input: {X[i]}, Expected: {y[i]}, Predicted: {prediction}, Error: {error}\nNew Weights: {weights}, New Bias: {bias:.2f}")

        print(f"Epoch {epoch + 1} Total Error: {total_error}\n")
        if total_error == 0:
            print("Perceptron learned the AND gate")
            break

    print("\nTraining Complete")
    print(f"Final Weights: {weights}, Final Bias: {bias:.2f}\n")


    print("Testing Perceptron for AND Gate")
    for i in range(len(X)):
        output = np.dot(X[i], weights) + bias
        prediction = activate(output) # Corrected line
        print(f"Input: {X[i]}, Predicted Output: {prediction}")

perceptron_and_gate(X, y)
```
